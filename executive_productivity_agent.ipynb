```python
# ------------------------------------------------------------
# DEADLINE URGENCY ENGINE
# ------------------------------------------------------------

def calculate_urgency(deadline, current_date):
    """
    Determines urgency based on the time remaining.
    """

    deadline = pd.Timestamp(deadline)
    current_date = pd.Timestamp(current_date)

    difference = deadline - current_date

    if difference.total_seconds() < 0:
        return "OVERDUE"

    hours = difference.total_seconds() / 3600

    if hours <= 24:
        return "CRITICAL"
    elif hours <= 48:
        return "HIGH"
    elif hours <= 72:
        return "MEDIUM"
    else:
        return "LOW"


tasks_df["urgency"] = tasks_df["deadline"].apply(
    lambda x: calculate_urgency(x, CURRENT_DATE)
)

tasks_df[
    ["task", "deadline", "status", "urgency"]
]
```

```python
# ------------------------------------------------------------
# TIME REMAINING
# ------------------------------------------------------------

def time_remaining(deadline, current_date):
    deadline = pd.Timestamp(deadline)
    current_date = pd.Timestamp(current_date)

    difference = deadline - current_date

    if difference.total_seconds() < 0:
        return "Overdue"

    total_hours = difference.total_seconds() / 3600

    if total_hours < 24:
        return f"{total_hours:.1f} hours"
    
    days = total_hours / 24

    if days < 2:
        return f"{days:.1f} days"
    
    return f"{days:.1f} days"


tasks_df["time_remaining"] = tasks_df["deadline"].apply(
    lambda x: time_remaining(x, CURRENT_DATE)
)

tasks_df[
    ["task", "deadline", "urgency", "time_remaining"]
]
```

```python
# ------------------------------------------------------------
# UNASSIGNED TASK DETECTOR
# ------------------------------------------------------------

def check_assignment(row):
    if row["owner"] == "Unassigned":
        return "ACTION REQUIRED"
    else:
        return "Assigned"


tasks_df["assignment_status"] = tasks_df.apply(
    check_assignment,
    axis=1
)

tasks_df[
    ["task", "owner", "assignment_status"]
]
```

```python
# ------------------------------------------------------------
# OPEN COMMITMENT DETECTOR
# ------------------------------------------------------------

def check_commitment(row):
    
    if row["owner"] == "Arjun Malhotra":
        
        completed_statuses = [
            "Confirmed",
            "Report received"
        ]
        
        if row["status"] not in completed_statuses:
            return "OPEN COMMITMENT"
    
    return "No open commitment"


tasks_df["commitment_status"] = tasks_df.apply(
    check_commitment,
    axis=1
)

tasks_df[
    ["task", "owner", "status", "commitment_status"]
]
```

```python
# ------------------------------------------------------------
# PRIORITY ENGINE
# ------------------------------------------------------------

def get_priorities(tasks_df, current_date):

    current_date = pd.Timestamp(current_date)

    priority_tasks = tasks_df[
        (tasks_df["urgency"].isin(["CRITICAL", "HIGH"]))
        |
        (tasks_df["assignment_status"] == "ACTION REQUIRED")
    ].copy()

    priority_tasks = priority_tasks.sort_values(
        by="deadline"
    )

    return priority_tasks


today_priorities = get_priorities(
    tasks_df,
    CURRENT_DATE
)

today_priorities[
    [
        "task",
        "urgency",
        "time_remaining",
        "status",
        "assignment_status"
    ]
]
```

```python
# ------------------------------------------------------------
# EXECUTIVE PRIORITY SUMMARY
# ------------------------------------------------------------

print("==============================================")
print("ARJUN'S PRIORITIES")
print("Date:", CURRENT_DATE.strftime("%A, %d %B %Y"))
print("==============================================")

for _, row in today_priorities.iterrows():
    
    print()
    print("Task:", row["task"])
    print("Urgency:", row["urgency"])
    print("Time remaining:", row["time_remaining"])
    print("Status:", row["status"])
    
    if row["assignment_status"] == "ACTION REQUIRED":
        print("⚠️ Action required: Ownership is unresolved.")
```

```python
# ------------------------------------------------------------
# EVIDENCE RETRIEVAL
# ------------------------------------------------------------

def get_evidence(task_name):
    
    matching_tasks = tasks_df[
        tasks_df["task"] == task_name
    ]
    
    if matching_tasks.empty:
        return "No evidence found."
    
    task = matching_tasks.iloc[0]
    
    return task["source"]


for _, row in today_priorities.iterrows():
    print()
    print("TASK:", row["task"])
    print("SOURCE:", get_evidence(row["task"]))
```
